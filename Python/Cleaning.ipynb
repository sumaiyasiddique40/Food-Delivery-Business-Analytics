```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
customers = pd.read_csv("customers.csv")
restaurants = pd.read_csv("restaurants.csv")
products = pd.read_csv("products.csv")
orders = pd.read_csv("orders.csv")
order_items = pd.read_csv("order_items.csv")
delivery = pd.read_csv("delivery.csv")
delivery_partners = pd.read_csv("delivery_partners.csv")
```

```python
tables = {
    "customers": customers,
    "restaurants": restaurants,
    "products": products,
    "orders": orders,
    "order_items": order_items,
    "delivery": delivery,
    "delivery_partners": delivery_partners
}

for name, df in tables.items():
    print(f"{name}: {df.shape}")
```

```python
for name, df in tables.items():
    print("\n" + "="*50)
    print(name)
    print(df.columns.tolist())
```

```python
for name, df in tables.items():
    print(
        name,
        "Duplicate rows:",
        df.duplicated().sum()
    )
```

```python
for name, df in tables.items():
    print("\n" + "="*50)
    print(name)
    print(df.dtypes)
```

```python
for name, df in tables.items():

    text_columns = df.select_dtypes(
        include="object"
    ).columns

    for col in text_columns:
        df[col] = df[col].str.strip()
```

```python
customers["signup_date"] = pd.to_datetime(
    customers["signup_date"],
    errors="coerce"
)

restaurants["joining_date"] = pd.to_datetime(
    restaurants["joining_date"],
    errors="coerce"
)

orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    errors="coerce"
)

delivery["pickup_time"] = pd.to_datetime(
    delivery["pickup_time"],
    errors="coerce"
)

delivery["delivery_time"] = pd.to_datetime(
    delivery["delivery_time"],
    errors="coerce"
)
```

```python
orders[
    (orders["subtotal"] < 0) |
    (orders["delivery_fee"] < 0) |
    (orders["discount"] < 0) |
    (orders["total_amount"] < 0)
]
```

```python
delivery[
    (delivery["actual_minutes"] < 0) |
    (delivery["estimated_minutes"] < 0)
]
```

```python
products[
    (products["price"] < 0) |
    (products["cost"] < 0)
]
```

```python
orders["order_month"] = (
    orders["order_date"]
    .dt.to_period("M")
    .astype(str)
)

orders["order_year"] = (
    orders["order_date"].dt.year
)

orders["order_day"] = (
    orders["order_date"].dt.day_name()
)
```

```python
delivery["delay_minutes"] = (
    delivery["actual_minutes"]
    - delivery["estimated_minutes"]
)

delivery["is_late"] = np.where(
    delivery["actual_minutes"] >
    delivery["estimated_minutes"],
    1,
    0
)
```

```python
for name, df in tables.items():

    print("\n", name)

    print(
        "Rows:",
        len(df)
    )

    print(
        "Missing values:",
        df.isnull().sum().sum()
    )

    print(
        "Duplicates:",
        df.duplicated().sum()
    )
```

```python
customers.to_csv(
    "customers_clean.csv",
    index=False
)

restaurants.to_csv(
    "restaurants_clean.csv",
    index=False
)

products.to_csv(
    "products_clean.csv",
    index=False
)

orders.to_csv(
    "orders_clean.csv",
    index=False
)

order_items.to_csv(
    "order_items_clean.csv",
    index=False
)

delivery.to_csv(
    "delivery_clean.csv",
    index=False
)

delivery_partners.to_csv(
    "delivery_partners_clean.csv",
    index=False
)
```
